### Exercise 10:
Write a Python function that takes two strings and checks if they are anagrams.

```python
# Exercise 10
```

````python
def are_anagrams(str1: str, str2: str) -> bool:
    
    clean_str1 = str1.replace(' ', '').lower()
    clean_str2 = str2.replace(' ', '').lower()

    if len(clean_str1) != len(clean_str2):
        return False

    return sorted(clean_str1) == sorted(clean_str2)

test_cases = [
    ("listen", "silent", "True Anagrams"),
    ("Debit Card", "Bad Credit", "True Anagrams with Case/Space"),
    ("hello", "olleh", "True Anagrams (simple)"),
    ("A Gentleman", "Elegant Man", "True Anagrams (complex)"),
    ("python", "java", "Different Lengths"),
    ("apple", "apply", "Different Letters"),
    ("elbow", "below", "True Anagrams"),
    ("race", "care.", "False - punctuation is not removed"),
]

print("--- Anagram Checker Results ---")
for s1, s2, description in test_cases:
    result = are_anagrams(s1, s2)
    print(f"Test: {description}")
    print(f"  String 1: '{s1}'")
    print(f"  String 2: '{s2}'")
    print(f"  Result: {'ok, Anagrams' if result else 'No, Not Anagrams'}\n")
print("-------------------------------")
````

### Exercise 11:
Write a Python program that checks if a number is prime.

```python
# Exercise 11
```

````python
import math

def is_prime(n: int) -> bool:


    if n <= 1:
        return False

    if n == 2:
        return True

    if n % 2 == 0:
        return False
    limit = int(math.sqrt(n)) + 1
    for i in range(3, limit, 2):
        if n % i == 0:
            return False

    return True


test_numbers = [
    1, 2, 3, 4, 11, 25, 49, 97, 100, 101, 121
]

print("--- Prime Number Check Results ---")
for number in test_numbers:
    result = is_prime(number)
    print(f"Is {number} prime? {' Yes' if result else ' No'}")
print("----------------------------------")
````

### Exercise 23:
Write a Python function that checks if a number is greater than its reverse.

```python
# Exercise 23
```

````python
def is_greater_than_reverse(number: int) -> bool:

    if number < 0:

        return False

    num_str = str(number)

    reversed_num_str = num_str[::-1]

    reversed_number = int(reversed_num_str)

    return number > reversed_number

def print_comparison_result(num: int):
    """A helper function to print the result clearly."""
    result = is_greater_than_reverse(num)
    reversed_num = int(str(num)[::-1]) if num >= 0 else 0

    status = "Yes" if result else "No"

    if num >= 0:
        print(f"Number: {num:<5} | Reverse: {reversed_num:<5} | Is {num} > {reversed_num}? {status}")
    else:
        print(f"Number: {num:<5} | Reverse check skipped (negative input).")

````

````python
def contains_letters_and_numbers(password: str) -> bool:

    if not password:
        return False

    has_letter = False
    has_digit = False
    for char in password:
        if char.isalpha():
            has_letter = True
        elif char.isdigit():
            has_digit = True
            if has_letter and has_digit:
                return True

    return has_letter and has_digit

````

````python
def is_between(target: float, lower_bound: float, upper_bound: float, inclusive: bool = True) -> bool:

    low = min(lower_bound, upper_bound)
    high = max(lower_bound, upper_bound)

    if inclusive:
        return low <= target <= high
    else:
        return low < target < high

number_to_check = 15
lower = 10
upper = 20
result_1 = is_between(number_to_check, lower, upper)
print(f"Is {number_to_check} between {lower} and {upper} (inclusive)? {result_1}")

number_on_boundary = 20
result_2 = is_between(number_on_boundary, lower, upper)
print(f"Is {number_on_boundary} between {lower} and {upper} (inclusive)? {result_2}")

number_exclusive = 10
result_3 = is_between(number_exclusive, lower, upper, inclusive=False)
print(f"Is {number_exclusive} between {lower} and {upper} (exclusive)? {result_3}")

number_inside_exclusive = 15
result_4 = is_between(number_inside_exclusive, lower, upper, inclusive=False)
print(f"Is {number_inside_exclusive} between {lower} and {upper} (exclusive)? {result_4}")

a = 100
b = 50
c = 75
result_5 = is_between(c, a, b)
print(f"Is {c} between {a} and {b} (inclusive)? {result_5}")

d = 5
result_6 = is_between(d, lower, upper)
print(f"Is {d} between {lower} and {upper} (inclusive)? {result_6}")
````

### Exercise 27:
Write a Python program that checks if a number is a perfect square.

```python
# Exercise 27
```

````python
import math

def is_perfect_square(n: int) -> bool:

    if n < 0:
        return False

    if n == 0:
        return True

    root = math.sqrt(n)

    return root % 1 == 0

print("--- Perfect Square Checks ---")

numbers_to_check = [9, 16, 25, 144, 0, 1]
for num in numbers_to_check:
    print(f"Is {num} a perfect square? {is_perfect_square(num)}")

print("-" * 30)

numbers_to_check_non = [8, 10, 26, 15, -4]
for num in numbers_to_check_non:
    print(f"Is {num} a perfect square? {is_perfect_square(num)}")

print("-" * 30)

large_square = 788481
print(f"Is {large_square} a perfect square? {is_perfect_square(large_square)}")
````
