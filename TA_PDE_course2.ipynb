```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
alpha = .05
x0 = np.linspace(-10,10,200)
def a(x): return 1 + .5*np.sin(x)

dt, t_final = .1, 5
x = np.copy(x0)
for _ in range(int(t_final/dt)):
  x = x + dt* a(x)
u0 = np.exp(-.5*(x0/1.5)**2)
u = u0*np.exp(-alpha*t_final)
plt.plot(x0, u0, label="t=0")
plt.plot(x,u, label=f"t={t_final}")
plt.legend()
plt.title("Nonuniform transport with decay")
plt.show()
```
