### Exercise 34:
Write a Python script to count the frequency of each letter in a text file `"message.txt"`.

```python
# Exercise 34
```

````python
try:
    with open("message.txt", "r") as fichier:
        texte = fichier.read().lower()
    
    frequence = {}
    for caractere in texte:
        if caractere.isalpha():
            frequence[caractere] = frequence.get(caractere, 0) + 1
    
    for lettre, count in sorted(frequence.items()):
        print(f"{lettre}: {count}")
        
except FileNotFoundError:
    print("Le fichier message.txt n'existe pas")
````

### Exercise 37:
Write a Python function that reads a file and removes all punctuation marks from the text.

```python
# Exercise 37
```

````python
import string

def supprimer_ponctuation(nom_fichier):
    try:
        with open(nom_fichier, "r") as fichier:
            texte = fichier.read()
        
        texte_sans_ponctuation = texte.translate(str.maketrans('', '', string.punctuation))
        
        with open(nom_fichier, "w") as fichier:
            fichier.write(texte_sans_ponctuation)
            
    except FileNotFoundError:
        print("Fichier non trouvé")

supprimer_ponctuation("texte.txt")
````

### Exercise 39:
Write a Python script to count the number of lines that start with a specific letter (e.g., "A") in a text file.

```python
# Exercise 39
```

````python
def compter_lignes_debut(nom_fichier, lettre):
    try:
        compteur = 0
        with open(nom_fichier, "r") as fichier:
            for ligne in fichier:
                if ligne.strip().startswith(lettre.upper()) or ligne.strip().startswith(lettre.lower()):
                    compteur += 1
        return compteur
    except FileNotFoundError:
        return 0

resultat = compter_lignes_debut("fichier.txt", "A")
print(resultat)
````

### Exercise 49:
Write a Python function that reads a file and returns the most common letter in the text.

```python
# Exercise 49
```

````python
def lettre_plus_commune(nom_fichier):
    try:
        with open(nom_fichier, "r") as fichier:
            texte = fichier.read().lower()
        
        frequence = {}
        for caractere in texte:
            if caractere.isalpha():
                frequence[caractere] = frequence.get(caractere, 0) + 1
        
        if frequence:
            return max(frequence, key=frequence.get)
        else:
            return None
            
    except FileNotFoundError:
        return None

resultat = lettre_plus_commune("texte.txt")
print(resultat)
````
